```python
#Check if its a right angled triangle
def is_right_angled(a,b,c):
    sides=sorted([a,b,c])
    x=sides[0]
    y=sides[1]
    z=sides[2]
    if z*z==x*x+y*y:
        return True
    else:
        return False
a=float(input("enter side 1"))
b=float(input("enter side 2"))
c=float(input("enter side 3"))

if is_right_angled(a,b,c):
    print("right_angled triangle")
else:
    print("not a right_angled triangle")
    
```
